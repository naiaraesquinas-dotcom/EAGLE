```python
import matplotlib.pyplot as plt
import cartopy.crs as ccrs
import cartopy.feature as cfeature

fig, ax = plt.subplots(
    figsize=(10, 8),
    subplot_kw={"projection": ccrs.PlateCarree()}
)

# Fondo geográfico
ax.add_feature(cfeature.LAND)
ax.add_feature(cfeature.COASTLINE)
ax.add_feature(cfeature.BORDERS)

# Red PyPSA
n.plot(
    ax=ax,
    bus_sizes=0.003,
    line_widths=1.5
)

# Encuadre aproximado de Bélgica
ax.set_extent([2.2, 6.7, 49.3, 51.8])

plt.show()
```

```python
import matplotlib.pyplot as plt
import cartopy.crs as ccrs
import cartopy.io.img_tiles as cimgt

# Mapa base de OpenStreetMap
osm = cimgt.OSM()

fig = plt.figure(figsize=(12, 9))
ax = plt.axes(projection=osm.crs)

# Zona que queremos mostrar
ax.set_extent(
    [2.2, 6.7, 49.3, 51.8],
    crs=ccrs.PlateCarree()
)

# Fondo real de OpenStreetMap
ax.add_image(osm, 8)

# -------------------------
# DIBUJAR LAS LÍNEAS
# -------------------------

for nombre_linea, linea in n.lines.iterrows():

    bus0 = n.buses.loc[linea.bus0]
    bus1 = n.buses.loc[linea.bus1]

    ax.plot(
        [bus0.x, bus1.x],
        [bus0.y, bus1.y],
        linewidth=2,
        transform=ccrs.PlateCarree()
    )

# -------------------------
# DIBUJAR LOS BUSES
# -------------------------

ax.scatter(
    n.buses.x,
    n.buses.y,
    s=80,
    zorder=5,
    transform=ccrs.PlateCarree()
)

plt.show()
```

```python
import matplotlib.pyplot as plt
import cartopy.crs as ccrs
import cartopy.feature as cfeature

fig = plt.figure(figsize=(12, 9))
ax = plt.axes(projection=ccrs.PlateCarree())

# Zona de Bélgica
ax.set_extent([2.2, 6.7, 49.3, 51.8])

# Fondo limpio
ax.add_feature(cfeature.LAND)
ax.add_feature(cfeature.OCEAN)
ax.add_feature(cfeature.COASTLINE)
ax.add_feature(cfeature.BORDERS)

# Dibujar líneas
for nombre_linea, linea in n.lines.iterrows():

    bus0 = n.buses.loc[linea.bus0]
    bus1 = n.buses.loc[linea.bus1]

    ax.plot(
        [bus0.x, bus1.x],
        [bus0.y, bus1.y],
        linewidth=2,
        transform=ccrs.PlateCarree()
    )

# Dibujar buses
ax.scatter(
    n.buses.x,
    n.buses.y,
    s=80,
    zorder=5,
    transform=ccrs.PlateCarree()
)

plt.show()
```
